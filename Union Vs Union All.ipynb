### If i want to write in Spark Sql 1st Convert in to Table 

```python
Manager_Data1_df.createOrReplaceTempView("Manager_Data1_df_tbl")
Duplicate_Manager_df.createOrReplaceTempView("Duplicate_Manager_df_tbl")
```

### Write in Spark Sql

```python
spark.sql("""
          select * from Manager_Data1_df_tbl
          union
          select* from Duplicate_Manager_df_tbl

          """).count()
```
