### Multiple table Joins

```python
customer_data.join(
    sales_data,
    customer_data.customer_id == sales_data.customer_id,
    "inner"
).join(
    product_data,
    sales_data.product_id == product_data.id,
    "inner"
).select(
    customer_data.customer_name,
    product_data.name,
    sales_data.quantity,
    product_data.price
).show()
```
