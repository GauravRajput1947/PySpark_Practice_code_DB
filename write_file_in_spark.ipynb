**First create dataFrame and **Read****

```python
df = spark.read.format("csv") \
    .option("header", "True") \
    .option("inferSchema", "true") \
    .load("/Volumes/workspace/default/write_spark/employee_write_data.csv")
```

```python
df.show()
```

```python
df.printSchema()
```

**Write DataFrame to Disk in spaRK**

```python
df.write.format("csv")\
    .option("header", "true")\
    .option("mode", "overwrite")\
    .option("path", "/Volumes/workspace/default/write_spark/csv_write")\
    .save(mode='overwrite')
```

```python
dbutils.fs.ls( "/Volumes/workspace/default/write_spark/csv_write")
```

## Using Repartition

```python
df.repartition(3).write.format("csv")\
    .option("header", "true")\
    .option("mode", "overwrite")\
    .option("path", "/Volumes/workspace/default/write_spark/csv_write_repartition")\
    .save()
```

```python
dbutils.fs.ls("/Volumes/workspace/default/write_spark/csv_write_repartition")
```

## Partitioning and Bucketing

**i want to create folder of different country**

```python
df = spark.read.format("csv") \
    .option("header", "true") \
    .load("/Volumes/workspace/default/write_spark/employee_write_data.csv")
df.show()

```

```python
df.write \
    .format("csv") \
    .mode("overwrite") \
    .option("header", "true") \
    .option("path", "/Volumes/workspace/default/write_spark/partition_by_address") \
    .partitionBy("address") \
    .save()
```

```python
dbutils.fs.ls("/Volumes/workspace/default/write_spark/partition_by_address")
```

## create partition on gender and same code use like address.

```python
df.write \
    .format("csv") \
    .mode("overwrite") \
    .option("header", "true") \
    .option("path", "/Volumes/workspace/default/write_spark/partition_by_address_gender") \
    .partitionBy("address","gender") \
    .save()
```

```python
dbutils.fs.ls("/Volumes/workspace/default/write_spark/partition_by_address_gender")
```

```python
dbutils.fs.ls("/Volumes/workspace/default/write_spark/partition_by_address_gender/address=INDIA/")
```

## Bucketing by

```python
df.write \
    .format("csv") \
    .mode("overwrite") \
    .option("header", "true") \
    .option("path","/Volumes/workspace/default/write_spark/bucket_by_id") \
    .bucketBy(3,"id") \
    .saveAsTable("bucket_by_id_table")
```
