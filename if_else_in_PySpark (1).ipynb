```python
Emp_df = (
    spark.read.format("csv")\
    .option("header", "false")\
    .option("inferSchema", "true")\
    .option("mode", "FAILFAST")\
    .load("/Volumes/workspace/default/employee_data/Emp_df.csv")
)
```

**Create Schema**

```python
emp_data = [
(1,'manish',26,20000,'india','IT'),
(2,'rahul',None,40000,'germany','engineering'),
(3,'pawan',12,60000,'india','sales'),
(4,'roshini',44,None,'uk','engineering'),
(5,'raushan',35,70000,'india','sales'),
(6,None,29,200000,'uk','IT'),
(7,'adam',37,65000,'us','IT'),
(8,'chris',16,40000,'us','sales'),
(None,None,None,None,None,None),
(7,'adam',37,65000,'us','IT')
]
schema = ['id','name','age','sal','country','dept']
emp_df=spark.createDataFrame(data=emp_data,schema=schema)
```

```python
emp_df.show()
emp_df.count()
```

```python
from pyspark.sql.functions import*
from pyspark.sql.types import*
```

```python
emp_df.createOrReplaceTempView("table")
```

**we use When otherwise and Case when in If-else in pyspark.**

```python
# age is greater than 18 you are Adult otherwise you are minor 

emp_df.withColumn("adult",when(col("age")<18,"NO")
                  .when(col("age")>18,"YES")
                  .otherwise("Novalue")).show()
```

```python
# suppose Age is null ? what should i do ?
# first fix the null value and then use case when statement 

emp_df.withColumn("age",when(col("age").isNull(),lit(19))
                        .otherwise(col("age")))\
                        .withColumn("adult",when(col("age")>18,"YES")
                                    .otherwise("NO")).show()
                                   
```

```python
emp_df.withColumn("age_wish",
     when((col("age") > 0) & (col("age") < 18), "Minor")
    .when((col("age") >= 18) & (col("age") < 30), "mid")
    .otherwise("Major")
).show()
     
```

```python
spark.sql("""
select *,
       case 
           when age < 18 then 'minor'
           when age > 18 then 'major'
           else 'novalue'
       end as adult
from table
""").show()
```
