## Transformation operation

```python
###### Import  #####
from pyspark.sql.functions import*
from pyspark.sql.types import*
from pyspark.sql.functions import col


###### Reading Data #####
employee_df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .option("mode", "permisive") \
    .load("/Volumes/workspace/default/my_volume1/employee.csv")

employee_df.show()


###### Printing Schema #####
employee_df.printSchema()

###### Creating Temporary View  #####
employee_df.createOrReplaceTempView("employee_tbl")
```

## Alisasing

```python
employee_df.select(col("id").alias("employee_id"),"name","age").show()
```

### # filter/where

```python
#### whose person salalry is more than 1.5 lakh ####
employee_df.filter(col("salary")>150000).show()

```

```python
#### age less than 18 but earn 1.5lakh ####
employee_df.filter(
    (col("salary")>150000) & (col("age") <18)).show()
```

### Literal(I want to pass some value in column)



```python
#####Ex:suppose , whenever fill form in anywhere they asked last name etc, then in made dafault value###
employee_df.select("*",lit("kumar").alias("last_name")).show()
```

### adding column

```python
employee_df.withColumn("sur_name",lit("singh")).show()
```

### Casting Data Type

```python
employee_df.withColumn("id",col("id").cast("string")).printSchema()
```

```python
employee_df.withColumn("salary",col("salary").cast("long")).printSchema()
```

### Removing Column

```python
employee_df.drop("id",col("name")).show()
```

## Spark Sql

```python
### filter###
spark.sql("""
          select*from employee_tbl 
          where salary>150000
          and age<18
              
          """).show()
```

```python
### literal###
spark.sql("""
          select*,"kumar" as last_name from employee_tbl 
          where salary>150000
          and age<18
              
          """).show()
```

```python
### Adding Column###
spark.sql("""
          select*,"kumar" as last_name,concat(name,last_name)as full_name
          from employee_tbl 
          
              
          """).show()
```

```python
### reanaming column ###
spark.sql("""
          select*,"kumar" as last_name,concat(name,last_name)as full_name,id as employee_id
          from employee_tbl 
          
              
          """).show()
```

```python
### casting DataType ###
spark.sql("""
          select*,"kumar" as last_name,concat(name,last_name)as full_name,cast(id as string)
          from employee_tbl 
          
              
          """).printSchema()
```
